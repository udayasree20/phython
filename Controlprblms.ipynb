```python
a = float(input("Enter first number:"))
b = float(input("Enter first number:"))
operator = input("Enter operator (+, -, *, /):")
if operator == "+":
    print("Result:", a+b)
elif operator == "-":
    print("Result:", a-b) 
elif operator == "*":
    print("Result:", a*b)
elif operator == "/":
    if b != 0:
       print("Result:", a/b) 
    else:
        print("cannot divide by 0")
else:
    print("invalid operator")
```
