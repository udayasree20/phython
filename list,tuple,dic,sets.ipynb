```python

#tuples are immutable, meaning they cannot be changed after they are created. They are defined
numbers = (10 ,20 , 20 , 30 , 20)

print(numbers.count(20))
```

```python
#index
numbers = (10 , 20 , 30 , 40)

print(numbers.index(30))
```

```python
numbers = (10 , 20 , 30 , 40)


print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```

```python
#sets in python 
#set is a collection of unique values that is unordered and mutable 
numbers = {10 , 20 , 30 , 20 , 10}

print(numbers)
```

```python
#add values to a set
subjects = {"python" , "java"}

subjects.add("SQL")

print(subjects)
```

```python
#remove values from a set
subjects = {"python" ,"SQL", "java"}
subjects.remove("java")

print(subjects)
```

```python
# sets do not allow duplicate values
numbers = {1 , 2 , 2 , 3 , 3 , 4}

print(numbers)
```

```python
#dictionaries in python
#dictionary is a collecton of key-value pairs that isunordered and mutable
student = {
    "name" : "uday",
    "age" : 17,
    "course"  : "python"

}
print(student)
```

```python
#acess elements in dictionary
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to a dictionary
student["city"] = "Tirupathi"

print(student)
```

```python
# Slicing 
numbers = [10 , 20 , 30 , 40 , 50 , 60 , 70 , 80]

print(numbers[-2:1:-3])
print(numbers[1:6:2])
print(numbers[::-4])
```

```python
student = {
    "name" : "uday",
    "age" : 17,
    "course"  : "python"
}

print(student.keys())
# key() returns all key-value pairs

print(student.values())
#values() return all values in dictionary

print(student.items())
# items() returns all key-value pairs

print(student.get("name"))
#get() returns the value of specified key

student.update({"age": 22 })
#update() updates the value of specified fee

print(student)

student.pop("age")
# pop() removes the specified key and its value

print(student)

```

```python
#pop item() removes the last inserted key-value pair
student = {
    "name" : "uday",
    "age" : 17,
    "course"  : "python"

}

student.popitem()

print(student)
```
