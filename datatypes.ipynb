```python

#list in python
#list is an ordered and changeable collection that can store multiple values.
marks = [80 , 90 , 75 , 85]

print(marks)
```

```python
#accessing elements in a list
marks = [80 , 90 , 75 , 85]

print(marks[0])  
print(marks[1])  
print(marks[2])

marks = [29 , 64 , 36 , 91 , 44 , 79]

print(marks[0])
print(marks[1])
print(marks[2])
print(marks[3])
print(marks[4])
print(marks[5])
```

```python
#change elements in a list
marks = [80 , 90 , 75]

marks[1] = 95
print(marks) 


marks = [79 , 64 , 92 , 74 , 96 , 77 , 94]

marks[2] = 100
marks[4] = 99
marks[6] = 88
print(marks)
```

```python
#add elements to a list 

marks = [80 , 90 , 75]
marks.append(85)
print(marks)
```

```python
#remove elements from a list
marks = [80 , 90 , 75]

marks.remove(90)
print(marks)
```

```python
#list of numbers
numbers = [10 , 20 , 30]

numbers.insert(1, 15)

print(numbers)
```

```python
#bitwise operators 
a=5
b=3

print(a & b)
print(a | b)
print(a ^ b)
print(a << b)
print(a >> b)
```

```python
#extend
a = [1, 2, 3]
b = [4, 5, 6]

a.extend(b)
print(a)
```

```python
#remove
numbers = [20, 20, 30, 20]
numbers.remove(20)
print(numbers)
```

```python
#clear
numbers = [10 , 20 , 30]
numbers.clear()
print(numbers)
```

```python
#index
numbers = [10, 20 , 30 , 40]
print(numbers.index(30))
```

```python
#count
numbers = [10, 20, 20, 30, 20]
print(numbers.count(20))
```

```python
#sort
numbers = [40,10,30,20]
numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)
```

```python
#reverse 
numbers= [10,20,30,40]
numbers.reverse()
print(numbers)
```

```python
#copy
a = [1 ,2 ,3]
b = a.copy()
print(b)
```

```python
numbers = [10 , 20 , 30 , 40 , 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1])
print(numbers[-4::])
print(numbers[-1::])
print(numbers[::-2])
```
